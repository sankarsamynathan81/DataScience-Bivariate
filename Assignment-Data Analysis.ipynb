# Q7)Test the Analysis of Variance between etest_p and mba_p at signifance level 5%.(Make decision using Hypothesis Testing)

```python
Notes:
    Hypothesis
        1. Null Hypothesis (H0) : There is no significant difference between the means of etest_p and mba_p.
        2. Alternative Hypothesis (H1) : There is a significant difference between the means of etest_p and mba_p
```

```python
import scipy.stats as stats

stats.f_oneway(dataset['etest_p'],dataset['mba_p'])
```

# Explanation:
    Desicion Rule:
        if p-value < 0.05 --> reject H0
        if p-value >=0.05 --> accept H0
        since p-value < 0.05, reject the null hypothesis. (result=4.672*10 power of -21, so 0.000000000000000000000046)
        There is a statiscally significant difference between etest_p and mba_p at the 5% significance level.
        

# Q8)Test the similarity between the degree_t(Sci&Tech) and specialisation(Mkt&HR) with respect to salary at significance level of 5%.(Make decision using Hypothesis Testing)

```python
Notes:
    Hypothesis
        1. Null Hypothesis (H0) : There is no significant difference in salary between students belonging to Sci&Tech degree and Mkt&HR specialization.
        2. Alternative Hypothesis (H1) : There is a significant difference in salary between the means these groups.
```

```python
from scipy.stats import ttest_ind
group1=dataset[dataset["degree_t"]=="Sci&Tech"]["salary"]
group2=dataset[dataset["specialisation"]=="Mkt&HR"]["salary"]
```

```python
group1
```

```python
stats.f_oneway(group1,group2) #this is oneway classification answer
```

```python
##t_statistic, p_values = ttest_ind(group1,group2,equal_var=False)#this is another way , but i am not sure which one we need to use here
t_statistic, p_values = ttest_ind(group1,group2)
```

```python
print("T_statistic",t_statistic)
print("P_values",p_values)
```

# Explanation:
    Decision Rule:
        If P-value < 0.05 --> Reject H0
        If P-value >=0.05 --> Accept H0
        If the obtained p-value is less than 0.05, reject the null hypothesis.
        otherwise, accept the null hypothesis.

# Q10)What is the probability Density Function of the salary range from 700000 to 900000?

```python
# Density Polt
import seaborn as sns
x=dataset['salary']
ax = sns.distplot(dataset['salary'],kde=True)
```

```python
def get_pdf_probability(dataset,startrange,endrange):
    from matplotlib import pyplot
    from scipy.stats import norm
    import seaborn as sns
    ax = sns.distplot(dataset,kde=True,kde_kws={'color':'blue'},color='Green')
    pyplot.axvline(startrange,color='Red')
    pyplot.axvline(endrange,color='Red')
    # generate a sample
    sample = dataset
    # calculate parameters
    sample_mean =sample.mean()
    sample_std = sample.std()
    print('Mean=%.3f, Standard Deviation=%.3f' % (sample_mean, sample_std))
    # define the distribution
    dist = norm(sample_mean, sample_std)
    
    # sample probabilities for a range of outcomes
    values = [value for value in range(startrange, endrange)]
    probabilities = [dist.pdf(value) for value in values]    
    prob=sum(probabilities)
    print("The area between range({},{}):{}".format(startrange,endrange,sum(probabilities)))
    return prob
    
```

```python
get_pdf_probability(dataset['salary'],700000,900000)
```

```python
salary1 = dataset[dataset["salary"]!=0]["salary"] #anotherway without graph using CDF-Cumulative detribution function
salary1
```

```python
mean1=salary1.min()
std1=salary1.std()
print(mean1,std1)
```

```python
from scipy.stats import norm
probability=norm.cdf(900000,mean1,std1)-norm.cdf(700000,mean1,std1)
print("Probability:",probability)
```

# Explanation:
    -This calculates the probability of observing salaries between 7,00,000 and 9,00,000 assuming a normal distribution.
    -The probability of salary falling between 7,00,000 and 9,00,000 is approximately 0.00000538, indicating that such salaries are very rare.

# Question-11)Test the similarity between the degree_t(Sci&Tech)with respect to etest_p and mba_p at significance level of 5%.(Make decision using Hy pothesis Testing)

```python
Notes:
    Hypothesis
    1. Null Hypothesis(H0): There is no significant difference between etest_p and mba_p among Sci&Tech students.
    2. Alternative Hypothesis(H1): There is a significant difference between etest_p and mba_p among Sci&Tech students.
```

```python
from scipy.stats import ttest_ind
sci=dataset[dataset["degree_t"]=="Sci&Tech"]
etest=sci["etest_p"]
mba=sci["mba_p"]
t_stat,p_value=ttest_ind(etest,mba)
print("T-statistic:",t_stat)
print("P_value",p_value)
if p_value < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Accept Null Hypothesis")
```

# Explanation:
    A t-test is used to determine whether the means differ significantly.
    Based on the obtained p-value, the hypothesis decision was made at the 5% siggnificance level.
